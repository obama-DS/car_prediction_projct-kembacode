## C04 | T01: Prepare features without leaking the answer
Use these fixed changes of scale:

`x1 = weight_kg / 1000`, `x2 = engine_liters / 5`, `x3 = cylinders / 8`.

A 1,500 kg car becomes `x1 = 1.5`. This changes numerical scale, not the physical car. Apply the SAME rule during training and inference. MPG stays in its original unit; no target conversion is needed.

A model-ready example has two parts: `[features, target]`. For instance `[[1.0, 0.5, 0.5], 30.0]`. The inner list contains inputs; 30 is the correct answer.

**Predict:** if you change only a row's target, should its features change?

📝 **Report Q6:** why must every split, and every new car, use the same divisors?

```python
def make_features(row, mode):
    # T01: keep the target OUT of the input list.
    if mode == "baseline":
        return []
    if mode == "mass":
        return [row[0] / 1000]
    if mode == "curved":
        x = row[0] / 1000
        return [x, (x - 1.4) ** 2]
    if mode == "all":
        # Return mass/1000, engine_liters/5, cylinders/8, in that order.
        return [row[0] / 1000, row[1] / 5, row[2] / 8]

def make_examples(rows, mode):
    examples = []
    for row in rows:
        examples.append([make_features(row, mode), row[3]])
    return examples

assert make_features([1000, 2.5, 4, 999], "all") == [1.0, 0.5, 0.5]
assert make_features([1000, 2.5, 4, -999], "all") == [1.0, 0.5, 0.5]
assert make_features([1000, 2.5, 4, 999], "mass") == [1.0]
TRAIN_A = make_examples(TRAIN_DATA, "mass")
TRAIN_B = make_examples(TRAIN_DATA, "all")
VAL_A = make_examples(VAL_DATA, "mass")
VAL_B = make_examples(VAL_DATA, "all")
TRAIN_CURVED = make_examples(TRAIN_DATA, "curved")
VAL_CURVED = make_examples(VAL_DATA, "curved")
print("one model-ready example:", TRAIN_B[0])
print("T01 passed: changing the target did not change the features.")
print("Curved example:", TRAIN_CURVED[0])
assert make_features([1400, 2.5, 4, 999], "curved") == [1.4, 0.0]
print("Curved feature test passed.")
```

## C05 | T02: Write the prediction rule
Model A: `predicted_mpg = b + w1*x1`.

Model B: `predicted_mpg = b + w1*x1 + w2*x2 + w3*x3`.

Both are linear regression. Multiple inputs do not automatically make a neural network necessary.

Implement one function that supports any matching number of features and weights. These test weights are hand-picked arithmetic checks, NOT learned parameters.

📝 **Report Q7:** calculate the first test's 37 MPG by hand and show every term.

```python
def predict(features, weights, bias):
    if len(features) != len(weights):
        raise ValueError("features and weights must have the same length")
    # T02: start at bias, then add each feature times its matching weight.
    total = bias
    for i in range(len(features)):
        total += features[i] * weights[i]
    return total

assert predict([1.0, 0.5, 0.5], [-10.0, -2.0, -4.0], 50.0) == 37.0
assert predict([], [], 25.0) == 25.0
print("T02 passed.")
```

📝 **Report Q8:** write down your baseline value and explain why a model must beat it.

```python
def mean_target(examples):
    if not examples:
        raise ValueError("Need at least one example.")
    # T03: average the targets. Each example is [features, target].
    total = sum(example[1] for example in examples)
    return total / len(examples)

TOY = [[[1.0], 5.0], [[2.0], 7.0]]
assert mean_target(TOY) == 6.0
baseline_mpg = mean_target(TRAIN_A)
print("Training-mean baseline:", round(baseline_mpg, 3), "MPG")
print("Every car receives this same baseline prediction.")
```

## C07 | T04: Define training loss and a readable metric
For each example: `error = prediction - target`.

**MSE** = average of `error * error`. We minimize this during training.

**MAE** = average of `abs(error)`. We use it for validation and the final report.

For `TOY`, weights `[1]`, bias `1`: predictions are 2 and 3; targets are 5 and 7; errors are -3 and -4. Verify both checks manually.

📝 **Report Q9:** show both toy results by hand.

**Remember:** an MAE of 3 MPG is an average absolute error on a specified set. It is NOT "97% accuracy" or a guarantee that every error is below 3.

```python
def mean_squared_error(examples, weights, bias):
    if not examples:
        raise ValueError("Need at least one example.")
    # T04a: average squared prediction errors.
    squared_errors = []
    for features, target in examples:
        prediction = predict(features, weights, bias)
        error = prediction - target
        squared_errors.append(error ** 2)
    return sum(squared_errors) / len(squared_errors)

def mean_absolute_error(examples, weights, bias):
    if not examples:
        raise ValueError("Need at least one example.")
    # T04b: average absolute prediction errors.
    absolute_errors = []
    for features, target in examples:
        prediction = predict(features, weights, bias)
        error = prediction - target
        absolute_errors.append(abs(error))
    return sum(absolute_errors) / len(absolute_errors)

assert mean_squared_error(TOY, [1.0], 1.0) == 12.5
print("T04 passed. MSE is in MPG squared; MAE is in MPG.")
```

## C08 | T05: Calculate how the parameters affect loss
For weight `j`, average `2 * error * features[j]` across training examples. For the bias, average `2 * error`.

`dw[j]` is a derivative, not the change to add directly. Gradient descent SUBTRACTS `learning_rate * derivative`.

`[0.0] * len(weights)` creates one running total for each weight. `j` is a position in the feature and weight lists.

📝 **Report Q10–Q11:** explain why the weight gradient is -11, and why subtracting a negative value increases the weight.

```python
def gradients(examples, weights, bias):
    if not examples:
        raise ValueError("Need at least one example.")
    dw = [0.0] * len(weights)
    db = 0.0
    for features, target in examples:
        error = predict(features, weights, bias) - target
        # T05: for each j, add 2*error*features[j] to dw[j].
        #      Then add 2*error to db ONCE for this example (not inside the j loop).
        for j in range(len(weights)):
            dw[j] += 2 * error * features[j]
        db += 2 * error
    n = len(examples)
    for j in range(len(weights)):
        dw[j] = dw[j] / n
    return dw, db / n

assert gradients(TOY, [1.0], 1.0) == ([-11.0], -7.0)
# Two-feature check: db must get 2*error once per example, not once per feature.
assert gradients([[[1.0, 2.0], 5.0]], [0.0, 0.0], 1.0) == ([-8.0, -16.0], -8.0), \
    "Check db: add 2*error once per example, outside the j loop."
dw, db = gradients(TOY, [1.0], 1.0)
w_after = 1.0 - 0.1 * dw[0]
b_after = 1.0 - 0.1 * db
print("One manual update:", w_after, b_after)
assert abs(mean_squared_error(TOY, [w_after], b_after) - 1.325) < 1e-9
print("Toy MSE changed from 12.5 to 1.325.")
```

## C09 | T06: Train the mass-only model
Complete only the parameter updates in the provided training function. The weights start at zero; the bias starts at the training mean. This is a valid starting point for linear regression.

Compute ALL gradients from the old parameters before changing any of them. Keep the input examples fixed. The loss history includes the initial loss and the loss after each update: `epochs + 1` values.

📝 **Report Q12–Q13:** record the first and last loss and the learned numbers, and explain the weight in words. Lower training loss does not yet prove the model works on new cars.

**Checkpoint:** if training is unstable, inspect scaling and the subtraction sign before changing anything else.

```python
def train(examples, learning_rate=0.05, epochs=2000):
    if not examples:
        raise ValueError('Need at least one example.')
    if not isinstance(epochs, int) or epochs < 1:
        raise ValueError('epochs must be a positive integer.')
    if not math.isfinite(learning_rate) or learning_rate <= 0:
        raise ValueError('learning_rate must be positive and finite.')
    weights = [0.0] * len(examples[0][0])
    bias = mean_target(examples)
    history = [mean_squared_error(examples, weights, bias)]
    for epoch in range(epochs):
        # Calculate every gradient BEFORE changing any parameter.
        dw, db = gradients(examples, weights, bias)
        # T06: update every weight and the bias, using the gradients above.
        for j in range(len(weights)):
            weights[j] -= learning_rate * dw[j]
        bias -= learning_rate * db
        current_loss = mean_squared_error(examples, weights, bias)
        if not math.isfinite(current_loss) or current_loss > 1e12:
            raise ValueError('Training became unstable. Check scaling, sign and learning rate.')
        history.append(current_loss)
    return weights, bias, history

# Quick test uses toy examples, not the final test set.
toy_w, toy_b, toy_history = train(TOY, learning_rate=0.05, epochs=1000)
assert len(toy_history) == 1001

assert mean_absolute_error(TOY, toy_w, toy_b) < 0.01
print("T06 passed: the toy model learned.")

w_a, b_a, history_a = train(TRAIN_A, learning_rate=0.05, epochs=2000)
print("Model A weights:", w_a, "bias:", b_a)
print("Model A MSE:", history_a[0], "->", history_a[-1])
show_plot([], [("training MSE", [[i, v] for i, v in enumerate(history_a)])],
          "Model A: training progress", "Parameter updates", "MSE (MPG squared)")
low = min(row[0] for row in TRAIN_DATA)
high = max(row[0] for row in TRAIN_DATA)
line = [[x, predict([x/1000.0], w_a, b_a)] for x in [low, high]]
show_plot([[row[0], row[3]] for row in TRAIN_DATA], [("trained line", line)],
          "Model A: predictions and training data", "Vehicle mass (kg)", "Fuel economy (MPG)")
# Optional curved model: two mass-based features
w_curved, b_curved, history_curved = train(
    TRAIN_CURVED, learning_rate=0.05, epochs=2000
)

print("Curved model weights:", w_curved, "bias:", b_curved)
print("Curved model training MAE:",
      mean_absolute_error(TRAIN_CURVED, w_curved, b_curved))
print("Curved model validation MAE:",
      mean_absolute_error(VAL_CURVED, w_curved, b_curved))
print("Model A validation MAE:",
      mean_absolute_error(VAL_A, w_a, b_a))
```

## C14 | T07: Build the final predictor
Finish the inference section of the supplied safety wrapper. It takes ordinary numbers and returns `(estimate, warnings)`.

The checks reject invalid inputs and warn about ranges/counts not represented in training. A range warning is a programmed check, NOT a confidence estimate. Inside-range inputs are not automatically reliable either.

`*specifications` passes the three numbers in the tuple as three arguments; the remaining arguments are the frozen model and training rows. This is supplied test syntax, not a new modeling idea.

The checks below run two ordinary requests, one out-of-range request and one invalid request. They also stop with an error if `estimate_mpg` calls `train`.

📝 **Report Q20:** copy the four results and explain them.

```python
def estimate_mpg(weight_kg, engine_liters, cylinders, model, training_rows):
    """Return a prediction and warnings; this is inference, not training."""
    for name, value in [('weight_kg', weight_kg), ('engine_liters', engine_liters), ('cylinders', cylinders)]:
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            raise ValueError(name + ' must be a number.')
        if not math.isfinite(value) or value <= 0:
            raise ValueError(name + ' must be positive and finite.')
    if int(cylinders) != cylinders:
        raise ValueError('cylinders must be a whole number.')
    warnings = ['Historical-data experiment only; not a real-world fuel guarantee.']
    for col, name, value in [(0,'mass',weight_kg),(1,'engine size',engine_liters)]:
        values = [row[col] for row in training_rows]
        if value < min(values) or value > max(values):
            warnings.append(name + ' is outside the training range.')
    if cylinders not in {row[2] for row in training_rows}:
        warnings.append('This cylinder count was not in the training set.')
    # T07: build the features in the frozen mode, then use frozen weights/bias.
    # Do NOT call train here. No actual MPG is available for a new request.
    features = [weight_kg / 1000]
    estimate = predict(features, FROZEN_MODEL["weights"], FROZEN_MODEL["bias"])
    if not math.isfinite(estimate) or estimate <= 0:
        warnings.append('The model gave a nonphysical result; no usable estimate.')
        return None, warnings
    return estimate, warnings

# REQUIRED APP CHECKS
def run_app_checks():
    before = (FROZEN_MODEL["weights"][:], FROZEN_MODEL["bias"])
    for specifications in [(1200, 2.0, 4), (1800, 5.0, 8), (4000, 12.0, 16)]:
        estimate, warnings = estimate_mpg(*specifications, FROZEN_MODEL, TRAIN_DATA)
        print("Inputs:", specifications, "Estimated MPG:", None if estimate is None else round(estimate, 1))
        for warning in warnings:
            print("  ", warning)
    assert before == (FROZEN_MODEL["weights"], FROZEN_MODEL["bias"]), "Inference changed the frozen model."
    try:
        estimate_mpg(-1, 2, 4, FROZEN_MODEL, TRAIN_DATA)
    except ValueError as error:
        print("Invalid-input check:", error)
    else:
        raise AssertionError("Negative mass must be rejected.")

# Inference must not train. While the checks run, train is swapped for a function that stops with an error.
_real_train = train
def _no_train(*args, **kwargs):
    raise AssertionError("estimate_mpg must not call train. Use the frozen weights and bias.")
train = _no_train
try:
    run_app_checks()
finally:
    train = _real_train  # always put the real training function back
print("T07 passed: predictions used the frozen model without retraining.")
```
